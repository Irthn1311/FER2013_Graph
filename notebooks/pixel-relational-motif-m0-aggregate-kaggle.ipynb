# PGM M0 zero-fit aggregate
Loads all 11 fixed prediction artifacts, then and only then loads Public labels for the registered bootstrap and gate. Executes zero model fits.

```python
from pathlib import Path
import shutil, subprocess, sys
REPO='https://github.com/Irthn1311/FER2013_Graph.git'; SHA='4b0349be2600c2797fc8fc61a8a6a48ffbabdb37'
P=Path('/kaggle/working/FER2013_Graph_M0_AGGREGATE_WRAPPER')
if P.exists(): shutil.rmtree(P)
subprocess.run(['git','clone',REPO,str(P)],check=True); subprocess.run(['git','-C',str(P),'checkout','--detach',SHA],check=True)
input_root=Path('/kaggle/input')
substrates=list(input_root.rglob('m0_substrate_manifest.json'))
r1s=list(input_root.rglob('e0r_r1_results.npz'))
prediction_markers=list(input_root.rglob('m0_prediction_input_manifest.json'))
if len(substrates)!=1 or len(r1s)!=1 or len(prediction_markers)!=1: raise RuntimeError((substrates,r1s,prediction_markers))
subprocess.run([sys.executable,str(P/'notebooks/m0-aggregate-entry.py'),'--prediction-root',str(prediction_markers[0].parent),'--substrate-root',str(substrates[0].parent),'--r1-results',str(r1s[0])],check=True)
```
